```python
import numpy as np
import pandas as pd

#Calculate total expenses for each month.

df =pd.read_csv('monthly_expense_tracker_2025.csv')
amount  = df['Amount'].to_numpy()
total_expense = np.sum(amount)
average_expense = np.mean(amount)
print(f"Total Expense : ₹{total_expense:,.2f}")
print(f"Average Expense : ₹{average_expense:,.2f}")


#Find the yearly total expense.

df['Date']=pd.to_datetime(df['Date'])
start_date = '2025-03-01'
end_date = '2025-06-30'
mask = (df['Date'] >= start_date) & (df['Date'] <= end_date)
filtered_df = df[mask]
total_expense = filtered_df['Amount'].sum()
catagory_summery = filtered_df.groupby('Category')['Amount'].sum()
print("\nCategory Summary:")
for category, amount in catagory_summery.items():
    print(f"{category}: ₹{amount:,.2f}")


#Find the average monthly expense.
monthly_totals = df.groupby('Month_Year')['Amount'].sum().reset_index()
monthly_total = filtered_df.groupby('Date')['Amount'].sum().reset_index()
monthly_totals['Date_Sort'] = pd.to_datetime(monthly_totals['Month_Year'], format='%B %Y')
monthly_totals = monthly_totals.sort_values(by='Date_Sort').drop(columns=['Date_Sort'])
avg_monthly_expense = monthly_totals['Amount'].mean()
median_monthly_expense = monthly_totals['Amount'].median()
min_monthly_expense = monthly_totals['Amount'].min()
max_monthly_expense = monthly_totals['Amount'].max()

print(f"Average Monthly Expense: ${avg_monthly_expense:,.2f}")
print(f"Median Monthly Expense: ${median_monthly_expense:,.2f}")
print(f"Min Monthly Expense: ${min_monthly_expense:,.2f}")
print(f"Max Monthly Expense: ${max_monthly_expense:,.2f}")
print("\nMonthly Totals Breakdown:")
print(monthly_totals.to_string(index=False))
```
